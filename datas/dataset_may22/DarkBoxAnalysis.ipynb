# Random, unfinshed modelling

By adjusting the LED width (w) and Voltage (V), we're adjusting the number of 260 nm photon $n_{260}(V,w)$ in a pulse. the explicit number of photon can be estimate by the power deliver to the LED times the width of the pulse, and divide by the energy of a 260 nm photon.

When 260 nm photon travels to the setup, it first interact with the material inside the container, few things happen at the same time.
1. the number of photon will be attenuate(Suppose it is ideal, only absorption happen, if not angle need to be accounted) by the material, the chance is given by $P_{ab}(\lambda=260\text{ nm},\text{mat},x_0)=e^{-\mu_{260,\text{mat}}x_0}$, here mat determine by the material.
2. if the photon absorbed by the material, it either
   1. Produce light at detectable (SiPM sense) range, so the number of scintillation light at wavelength $\lambda$ will be $$n_{\text{scint}}(\lambda,x_0)=n_{260}(V,w)P_{ab}(260,\text{mat},x_0)P_{\text{scint}}(\lambda)$$ where this $P_{\text{scint}}$ determined by the emission spectrum of the scintillator.
      1. Note that it is possible the emission light is of the absorption region of the material itself... We neglect that for simplicity
   2. Just disappear from the setup
3. Further attenuate by the thin glass layer (just neglect...)
4. After the light leaving the vials, we expect to have
   1. $n_{260,leave}=n_{260}(V,w)P_{ab}(260,\text{mat})$
   2. $n_{\lambda/260,leave}=n_{260}(V,w)P_{ab}(260,\text{mat})P_{\text{scint}}(\lambda)$

e.g. In the measurement of water, with known $n_{260}$, we are measuring $n_{260,leave}$. The same case we measure Quantum dots, we're measure directly $n_{260,leave}+\sum_{\lambda}n_{\lambda/260,leave,\text{scint}}$
