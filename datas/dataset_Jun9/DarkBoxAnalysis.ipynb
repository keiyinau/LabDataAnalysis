```python
setup1=pd.read_csv("DarkBox/WithoutVeto.txt",sep=" ")
setup2=pd.read_csv("DarkBox/WithVeto.txt",sep=" ")
ch0_fp=pd.read_csv("DarkBox/ch0_205.txt",sep=" ")
ch1_fp=pd.read_csv("DarkBox/ch1_205.txt",sep=" ")
channel_consideration="ch1"
bins_num="auto"

plt.hist(setup1['ch0'], bins=bins_num,histtype='step',label="SiPM1 (Without Veto)")
plt.hist(setup2['ch0'], bins=bins_num,histtype='step',label="SiPM2 (With Veto)")

#plt.hist(setup2['ch1'], bins=bins_num,histtype='step',label="SiPM2, %s, count=%.2f"%(case,len(SiPM2['ch1'])))
#plt.hist(Pedestal['ch0'], bins=bins_num,label="Pedestal ch0")
#plt.hist(ch0_fp['ch0'], bins=bins_num,label="Pedestal ch0")
#plt.hist(ch1_fp['ch1'], bins=bins_num,label="Pedestal ch1")
plt.title("Veto vs without Veto")
plt.xlabel("QDC counts")
plt.ylabel("Counts")
plt.xlim(0,3000)
plt.ylim(0,10000)
plt.legend()
plt.show()
```
