```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import os
import mplhep as hep
hep.style.use(hep.style.ROOT)
```

# Measurement result

```python
setup1=pd.read_csv("DarkBox/MajorityWidthSetting/SiPM1SiPM2_Maj_30.txt",sep=" ")

bins_num="auto"

plt.hist(setup1['ch0'], bins=bins_num,histtype='step',label="SiPM1")
plt.hist(setup1['ch1'], bins=bins_num,histtype='step',label="SiPM2")


#plt.hist(ch0_fp['ch0'], bins=bins_num,label=Pedestal ch0)
#plt.hist(ch1_fp['ch1'], bins=bins_num,label=Pedestal ch1)
plt.title("Dark Box Measurement Maj_30ns")
plt.xlabel("QDC counts")
plt.ylabel("Counts")
plt.xlim(0,3000)
plt.ylim(0,10000)
plt.legend()
plt.show()
```
